```python
vk=dl[dl['batsman']=='V Kohli']
```

```python
vk.groupby('bowling_team')['batsman_runs'].sum().sort_values(ascending=False).head(3)
```

```python
def run_score(batsman_name):
    vk=[dl['batsman']==batsman_name]
    return vk.groupby('bowling_team')['batsman_runs'].sum().sort_values(ascending=False).head(1)
```

```python
run_score('MS Dhoni')
```

```python
do=dl[dl['over']>15]
```

```python
ab=do.groupby('batsman')['batsman_runs'].count()
```

```python
x=ab>200
bl=ab[x].index.tolist()
```

```python
final=do[do['batsman'].isin(bl)]
```

```python
runs=final.groupby('batsman')['batsman_runs'].sum()
```

```python
ball=final.groupby('batsman')['batsman_runs'].count()
```

```python
sr=(runs/ball)*100
```

```python
sr
```
