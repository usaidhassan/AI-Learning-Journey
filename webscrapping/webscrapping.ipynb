```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
import numpy as np
```

```python
webpage=requests.get('https://books.toscrape.com/catalogue/page-1.html').text

soup=BeautifulSoup(webpage,'lxml')
```

```python
soup.find_all('h1')[0].text
```

```python
books=soup.find_all('article',class_='product_pod')
```

```python
len(books)
```

```python
name=[]
price=[]
instock=[]

for i in books:

  name.append(i.find('h3').text.strip())
  price.append(i.find('p',class_='price_color').text.strip())
  instock.append(i.find('p',class_='instock').text.strip())
  

df=pd.DataFrame({'name':name,
   'price':price,
   'instock':instock,
   })

```

```python
final=pd.DataFrame()
for j in range(1,51):
    webpage=requests.get(f'https://books.toscrape.com/catalogue/page-{j}.html').text

    soup=BeautifulSoup(webpage,'lxml')
    books=soup.find_all('article',class_='product_pod')
    name=[]
    price=[]
    instock=[]
    name=[]
    price=[]
    instock=[]

    for i in books:

        name.append(i.find('h3').text.strip())
        price.append(i.find('p',class_='price_color').text.strip())
        instock.append(i.find('p',class_='instock').text.strip())
  

    df=pd.DataFrame({'name':name,
   'price':price,
   'instock':instock,
   })
    final = pd.concat([final, df], ignore_index=True)

```
