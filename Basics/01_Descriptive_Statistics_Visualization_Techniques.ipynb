# Working with single Data/Variable
## 1. working with Categorical Data
### a. Frequency 

```python
data.columns
```

```python
data['Product_Category'].value_counts()
```

### b. Bar Chart

```python
data['Gender'].value_counts().plot(kind='bar')
```

### c. Pie Chart

```python
data['Gender'].value_counts().plot(kind='pie', autopct="%0.2f" )
```

### d. Cumulative Frequency

```python
a= data['Gender'].value_counts()
```

```python
a
```

```python
b = ((a/a.sum())*100).values
```

```python
b
```

```python
c=[]
cf=0
for i in b:
    cf=cf+i
    c.append(cf)
```

```python
plt.plot(c, marker = "D")
```
