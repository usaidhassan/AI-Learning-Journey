```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data=pd.read_csv('Iris.csv')
data.head()
```

```python
data.rename(columns={'SepalLengthCm':'SL','SepalWidthCm':'SW','PetalLengthCm':'PL','PetalWidthCm':'PW'},inplace=True)
```

```python
iris_setosa=data[data['Species']=='Iris-setosa']
iris_versicolor=data[data['Species']=='Iris-versicolor']
iris_virginica=data[data['Species']=='Iris-virginica']
```

```python
sns.histplot(iris_setosa['PL'], kde=True)
```

```python
sns.FacetGrid(data, hue='Species', height=6).map(sns.histplot,'PL', kde=True).add_legend()
```

```python
sns.FacetGrid(data, hue='Species', height=6).map(sns.histplot,'PW', kde=True).add_legend()
```

```python
sns.FacetGrid(data, hue='Species', height=6).map(sns.histplot,'SL', kde=True).add_legend()
```

```python
sns.FacetGrid(data, hue='Species', height=6).map(sns.histplot,'SW', kde=True).add_legend()
```

```python
counts, bin_edges=np.histogram(iris_setosa['PL'], bins=10)
```

```python
pdf=counts/sum(counts)
cdf= np.cumsum(pdf)
```

```python
plt.plot(bin_edges[1:],pdf)
plt.plot(bin_edges[1:],cdf)
```

```python
plt.style.use('fivethirtyeight')
counts, bin_edges=np.histogram(iris_setosa['PL'], bins=10)
pdf=counts/sum(counts)
cdf= np.cumsum(pdf)
plt.plot(bin_edges[1:],pdf)
plt.plot(bin_edges[1:],cdf)

counts, bin_edges=np.histogram(iris_versicolor['PL'], bins=10)
pdf=counts/sum(counts)
cdf= np.cumsum(pdf)
plt.plot(bin_edges[1:],pdf)
plt.plot(bin_edges[1:],cdf)

counts, bin_edges=np.histogram(iris_virginica['PL'], bins=10)
pdf=counts/sum(counts)
cdf= np.cumsum(pdf)
plt.plot(bin_edges[1:],pdf)
plt.plot(bin_edges[1:],cdf)
```

From this Graph We have clear probality of dicing wheter its a versicolor or virgenica
